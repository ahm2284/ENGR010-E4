#### 1. Reading in the data

__Explanation:__ <br> To begin, we are creating a file that reads a DNA sequence. Essentially, there is another file here that is a text file and contains a DNA sequence. Our goal for this task is to create a function that reads that file and sorts through the information to remove any bad formatting, or extraneous characters, and then return this information as a list of characters. To do this, we created a function that takes in one parameter, which is the file name. Then we create a variable that has an empty list that we will be using later to append the new list. Following that, we open the file, using the parameter as the fileName, and then 'r' to read the file and we read the file as r. Then, we initialize a variable that has an empty string for each line to hold the sorted-through data. After, we use another for loop but this time for each character in the line, if the character is in the string 'ATCG', so any of the correct DNA letters, then that character will be added to that string. After this loop goes for every character in the line, it will append whatever was in that string to the list. This whole process then repeats for every line in the file. When that is complete, we use the .join method to join everything onto a single string and then return that as a list where each letter is an item in the list. Finally, we call the function using the text file and print the result. 

```python
def readDNA(fileDNA):
    sequence = []  
    with open(fileDNA, 'r') as file:
        for line in file:
            fileDNA = ''  
            for character in line:
                if character in 'ATCG':
                    fileDNA += character  
            sequence.append(fileDNA)  
    combinedSequence = ''.join(sequence)  
    return list(combinedSequence)  
print(readDNA("DNA_NM_1.txt"))
```

#### 2. Reading in the translation table

__Explanation:__ <br> For this part, we are going to be reading a file that has contents that are in the format of a dictionary and opening it here. To make it so that we can use this code with any other file that has a similar composition, we are going to make this a function that takes in one parameter which will be the name of the file. After defining the function, we are going to create a variable that is currently a blank dictionary but we will be adding keys and values to it later. Following that, we open the file, which we say is the parameter, and then 'r' to read the file as file. Then, to a new variable, we are going to clean up what is coming in from the file to make sure that it is all formatted correctly. First, we are going to start by removing all the whitespace with strip, then we are going to remove the single quotes and commas using the replace method. Then, we are going to split these cleaned lines at the spaces using the split method. Then we create another for loop inside of this loop so that for every pair in the new variable that has the pairs, we split the pair at the colon to separate the key and the value or in this case codon and amino acid. Then, if the length of the variable is 2, we check if it split correctly into 2 different parts. If that condition is fulfilled, then we separate the key and value since the split method splits the string from "ATA:I" to ["ATA" , "I"], we use the indices to set the variables equal to their respective set of information. So codon will equal the first thing in the list and aminoAcid will equal the second thing in the list because the strip method gets rid of the rest. After that, we use the variables we have to add to the dictionary with the codon as the key and the aminoAcid as the value and iterate through the loop until dictionary is filled. When that is complete, then we return dictionary and call the function using the file name provided. 

```python
def translationTable(filename):
    dictionary = {}
    with open(filename, 'r') as file:
        for line in file:
            cleanedLine = line.strip().replace("'", "").replace(",", "")
            pairs = cleanedLine.split()
            for pair in pairs:
                codonAmino = pair.split(":")
                if len(codonAmino) == 2:
                    codon = codonAmino[0].strip()
                    aminoAcid = codonAmino[1].strip()
                    dictionary[codon] = aminoAcid
    return dictionary
dict1 = translationTable('DNA_Amino_Acid_Translation_Matrix.txt')
print(dict1)
```

#### 3. Create the Amino Acid Sequence from the DNA sequence

__Explanation:__ <br> For this step, we are going to create a function that takes four parameters, the DNA sequence to be used, the translation guide to use, the start index, and the stop index to convert the DNA sequence to an amino acid sequence by using every three characters in the DNA as a key in the dictionary and the adding the value associated with that key to a new variable to print, all starting at the index provided and stopping at the index provided to ensure that everything converts smoothly. To start, we define the function with the four parameters. Since DNA is a large list and we want to narrow it down to whatever indices are provided with start and stop, we set dna equal to the information inside dna but from the indices [start:stop] but we make sure to subtract one because in biology they start counting at one, not zero as they do in python. Following that, we make a new variable chunks which is an empty list that will be filled with the DNA broken up into 3 characters which is a codon. Then we create a for loop that for i in the range 0 to the length of DNA, that skips every three where i will be the starting index of each codon, and inside, we append the dna from index i to index i + 3 which translates to a codon and appends that to the list chunks. After going through all the data in the file, we create another empty string that will contain the translated string. We then create another for loop for each chunk in the list chunks, if the chunk is in the dictionary that we created earlier, then it will append the value associated with that chunk as chunk is the key and we are adding the value to the string. We continue to do that until we go through all the chunks and then return the string that contains the translated list. To call the function, we first have to call the readDNA function from above. Once we do that, we store that value into a variable but since that variable is a list and in this function we need it as a string, we use the .join method to turn that list into a string and set that into another variable. Since we also need to use the translationTable function from above, we call the translationTable function with the designated file and set that dictionary equal to a variable. Now that we have all of our parameter values, we call the function with each of the variables we created and the start and stop values and print that variable with the results of the DNA sequence to amino acid sequence. 

```python
def DNAtoAmino(dna, translation, start, stop):
    dna = dna[start-1: stop-1]
    chunks = []
    for i in range(0, len(dna), 3):
        chunks.append(dna[i:i+3])
        aminoResults = ''
        for chunk in chunks:
            if chunk in translation:
                aminoResults += translation[chunk]
    return aminoResults

dnaFileSequence = (readDNA("DNA_NM_1.txt"))
dnaFileString = ''.join(dnaFileSequence)

dict1 = translationTable('DNA_Amino_Acid_Translation_Matrix.txt')
result = DNAtoAmino(dnaFileString, dict1, 23, 1243)

print("dna file to amino acid with base numbers 23-1243 from protein file 1:", result)
```

#### 4. Create the Official Amino Acid Sequence

__Explanation:__ <br> For this, we need to just read a file and return the sequence as a list. This is similar to the function we created in the first part, but we do not need to clean up any of the data. We create a function that takes only one parameter which will be the file name. Then, we create an empty variable sequence that is currently an empty string. Then we open the file that is the parameter and use 'r' to read the file as file. Following that, for each line in the file, we are going to add the line stripped of any white space to whatever is in the variable sequence at the moment and continue iterating through that loop for all of the lines. Finally, we will return the string sequence turned into a list using list(sequence) and print the value of the function called after providing the file name. 

```python
def readAminoAcid(fileName):
    sequence = ''
    with open(fileName, 'r') as file:
        for line in file:
            sequence += line.strip() 
    return list(sequence)  
print(readAminoAcid("Protein_1_23_1243.txt"))
```

#### 5. Compare two Amino Acid Sequences

__Explanation:__ <br> Next, we are going to write a function that compares the value to two different sequences to see if they are equal. To do this, we are first going to create a small function that is the exact same function as the one in part 4, but instead of taking the extra step to return it as a list, we are just going to return the variable sequence to read the file as a string. Once we do that, we call this function with the protein file and set that result equal to a variable. Then, we call the function from part 3 that translates the DNA sequence and set that equal to another variable. Now that we have our variable, we created the function that will compare these two variables. We define the function with two parameters, for the two strings we are going to compare, and then create a new empty string variable that will print our result after comparing the two strings. We have it so that if the two strings equal each other, then the empty string will now equal True and if they do not equal each other, then it will equal False and return the string after running that. Finally, we print the result of the called function that compares the two strings with each of the variables with the value of the amino acids directly from the file and the amino acids translated from the DNA sequence as the two parameters. 

```python
def aminoAcidString(filename):
    sequence = ''
    with open(filename, 'r') as file:
        for line in file:
            sequence += line.strip()    
    return sequence
fileSequence = aminoAcidString("Protein_1_23_1243.txt")
result = DNAtoAmino(dnaFileString, dict1, 23, 1243)

def compareSequences(string1, string2):
    resultString = ''
    if string1 == string2:
        resultString = 'True'
    else:
        resultString = 'False'
    return resultString
print(compareSequences(fileSequence, result))
```
